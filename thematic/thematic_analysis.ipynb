# Thematic Analysis of Free-Text Reflections

**Paper:** *Bridging the AI Readiness Gap Among Clinicians: Results from a National, Interdisciplinary Health Datathon for Trainees*

**Produces:** Figures 3–4 (theme prevalence / sentiment / co-occurrence) and Supplementary Figure S1 (silhouette diagnostic).

---

## What this notebook does

Applies a codebook/consensus thematic analysis with computational code aggregation to
31 free-text team reflections collected at the close of the datathon.

Pipeline:
1. Load and filter reflections (31 collected → 21 usable after removing blank/duplicate/nonsense entries)
2. Load the dual-coder coding matrix (140 unique codes, 195 assignments; coders: KB, SM)
3. Cluster codes into themes via consensus silhouette analysis (k=7 selected on marginal-gain criterion)
4. Compute per-theme prevalence, sentiment (positive/neutral/negative), and co-occurrence
5. Generate Figures 3, 4, and Supplementary Figure S1

**Key results reproduced:**
- 7-theme prevalence: 16 / 11 / 8 / 4 / 4 / 3 / 3 (Teamwork, Interdisciplinary Learning,
  Motivation, Clinical Translation, Technical Skill, Communication, Equity)
- Strongest co-occurrences: Teamwork × Interdisciplinary = 10, Interdisciplinary × Motivation = 5,
  Teamwork × Motivation = 4
- Consensus silhouette: 0.27 at k=7, rising to 0.39 at k=12
