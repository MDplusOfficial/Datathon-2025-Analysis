# Team Diversity Analysis

**Paper:** *Bridging the AI Readiness Gap Among Clinicians: Results from a National, Interdisciplinary Health Datathon for Trainees*

**Produces:** Figure 1E (team composition and diversity metrics).

---

## What this notebook does

Analyzes team composition across the 49 registered teams (31 submitting) using
the team roster (`Manuscript_Teams.csv`).

Pipeline:
1. Load team roster and canonicalize institution names
2. Assign training levels to a closed degree ontology (MD, DO, PhD, MS, MPH, etc.)
3. Compute inter-disciplinary diversity (≥2 distinct training backgrounds per team)
4. Compute inter-institutional diversity (≥2 institutions per team)
5. Generate Figure 1E

**Key results reproduced:**
- 49 teams registered, 31 submitted (63%)
- 220 team participants total; 149 on submitting teams
- Interdisciplinary teams (≥2 backgrounds): 42.8% of registered, 44.9% of submitting
- 75 unique institutions represented
